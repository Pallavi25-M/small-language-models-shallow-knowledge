# GPT2

```python
from transformers import AutoTokenizer, AutoModelForCausalLM

tokenizer = AutoTokenizer.from_pretrained("openai-community/gpt2")
model = AutoModelForCausalLM.from_pretrained("openai-community/gpt2", device_map="auto")
```

```python
prompt = 'Is "car" an adverb, a noun, or a verb?'
```

```python
inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

# Generate text
outputs = model.generate(
    **inputs,
    max_new_tokens=100,
    do_sample=True,
    temperature=0.1,
    top_p=0.9
)

# Decode the generated tokens
generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)

print(generated_text)
```

```python
from transformers import pipeline

pipe = pipeline("text-generation", model="Qwen/Qwen3-0.6B", max_new_tokens=256)
```

```python
messages = [
    {
        "role": "system", "content": "Your job is to provide a \"yes\" or \"no\" answer. Nothing more.",
        "role": "user", "content": "Is \"car\" an adjective or a verb?"
    },
]
pipe(messages)
```
